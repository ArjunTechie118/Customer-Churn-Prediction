## **Comparing Models: RF & XGBoost**

- Because the target variable is imbalanced, accuracy alone is not a reliable metric for model selection. I focus more on recall, F1-score, precision, and ROC-AUC, with recall being particularly important because the goal is to identify customers who are likely to churn.

**I compared Random Forest and XGBoost using the same train-test split and the same preprocessing pipeline. Since the churn classes were imbalanced, I didn't rely on accuracy alone. XGBoost achieved slightly better recall, F1-score, and ROC-AUC than Random Forest, which were more relevant to our objective of identifying potential churners. Random Forest had slightly higher precision and accuracy, but overall I selected XGBoost because it provided a better balance between identifying churners and maintaining precision. I then chose XGBoost for further hyperparameter tuning.**

- Threshold and Trade-off:

*Where should I put the decision boundary so that the model's behavior matches my objective?*

```python

thresholds = [0.30, 0.35, 0.40, 0.45, 0.50, 0.55, 0.60]

results = []

for threshold in thresholds:
    
    Y_pred_threshold = (Y_prob_xgb_tuned >= threshold).astype(int)
    
    precision = precision_score(Y_test, Y_pred_threshold)
    recall = recall_score(Y_test, Y_pred_threshold)
    f1 = f1_score(Y_test, Y_pred_threshold)
    
    results.append({
        "Threshold": threshold,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1
    })

threshold_df = pd.DataFrame(results)

print(threshold_df)
```

- **After training the model, I analyzed different probability thresholds because the default 0.50 threshold was not giving the desired precision-recall balance. I selected 0.40 based on F1 performance and the need to improve churn recall.**

## **Temprory Model Selected**

Model: Tuned XGBoost

Threshold: 0.40


```python
def create_tenure_group(df):
    df = df.copy()

    df["Tenure Group"] = pd.cut(
        df["Tenure Months"],
        bins=[-1, 12, 24, 48, float("inf")],
        labels=["0-12", "13-24", "25-48", "49+"]
    )

    return df
```

```python
from sklearn.preprocessing import FunctionTransformer

tenure_transformer = FunctionTransformer(
    create_tenure_group,
    validate=False
)
```
